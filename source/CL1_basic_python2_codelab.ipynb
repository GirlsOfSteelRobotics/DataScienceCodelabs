```python

"""
A triangle is considered valid if the sum of two of the sides is smaller than the largest side.
"""

def is_valid_triangle(side1, side2, side3):
    sides = sorted([side1, side2, side3])
    xxx = (sides[0] + sides[1]) > sides[2]
    print(sides, xxx)
    return xxx


assert is_valid_triangle(1, 2, 3) == False
assert is_valid_triangle(3, 2, 1) == False
assert is_valid_triangle(2, 1, 3) == False
assert is_valid_triangle(3, 4, 5) == True
assert is_valid_triangle(12, 5, 13) == True
assert is_valid_triangle(8, 1, 4) == False
assert is_valid_triangle(100, 3504, 1) == False
```
